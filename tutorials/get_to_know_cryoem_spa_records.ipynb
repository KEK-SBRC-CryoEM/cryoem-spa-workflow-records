# Get to know the Cryo-EM SPA Workflow Records Dataset

This tutorial introduces the **Cryo-EM SPA Workflow Records** dataset. This dataset follows the "ArXiv" hierarchical structure where metadata is stored at the **Process ID** level (3rd Tier), allowing researchers to reconstruct processing steps without storing massive intermediate binary files like micrographs or particles.

## 1. Background
Instead of providing heavy `data/micrographs/` or `data/particles/` directories, this dataset provides the metadata required to re-run processing steps (e.g., Motion Correction, Particle Extraction) to reproduce those results.

### Data Hierarchy (ArXiv Tiering):
- **Tier 1: Project ID** (e.g., `PRJ001/`)
- **Tier 2: Dataset ID** (e.g., `DS001/`)
- **Tier 3: Process ID** (e.g., `PROC001/`) -> **Metadata files stored here.**

```python
import boto3
import json
import pandas as pd
from botocore import UNSIGNED
from botocore.config import Config

# Constants for the dataset
BUCKET_NAME = 'your-cryoem-open-data-bucket'
PROJECT_ID = 'example-project'
DATASET_ID = 'example-dataset'

s3 = boto3.client('s3', config=Config(signature_version=UNSIGNED))
```

## 2. Exploring Metadata (Tier 3)
We list the Process IDs to find the metadata JSON files.

```python
def list_process_ids(bucket, project, dataset):
    prefix = f"{project}/{dataset}/"
    paginator = s3.get_paginator('list_objects_v2')
    result = paginator.paginate(Bucket=bucket, Prefix=prefix, Delimiter='/')
    return [cp['Prefix'] for page in result for cp in page.get('CommonPrefixes', [])]

processes = list_process_ids(BUCKET_NAME, PROJECT_ID, DATASET_ID)
print(f"Processes found: {processes}")
```

## 3. Conclusion
By providing metadata at the **Process ID** level, we ensure reproducibility and efficiency.
